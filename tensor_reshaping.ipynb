```python
import torch
```

```python
# Tensor Reshaping

x = torch.arange(9)

# similar method
x_3x3 = x.view(3, 3) # view acts on contiguous tensors meaning the tensor is stored contiguously in memory (superior)
x_3x3 = x.reshape(3, 3) # safe bet (performance loss)
print(x_3x3)
```

```python
y = x_3x3.t() # tranpose the 3x3
print(y)
print(y.contiguous().view(9))
print(y.reshape(9))
```

```python
x1 = torch.rand((2, 5))
x2 = torch.rand((2, 5))
print(torch.cat((x1, x2), dim=0).shape)
print(torch.cat((x1, x2), dim=1).shape)
```

```python
z = x1.view(-1)
print(z.shape)
```

```python
batch = 64
x = torch.rand((batch, 2, 5))
z = x.view(batch, -1)
print(z.shape)
```

```python
z = x.permute(0, 2, 1)
print(z.shape)
```

```python
x = torch.arange(10)
print(x.unsqueeze(0).shape)
print(x.unsqueeze(1).shape)
```

```python
x = torch.arange(10).unsqueeze(0).unsqueeze(1) # 1x1x10
print(x.shape)
```
