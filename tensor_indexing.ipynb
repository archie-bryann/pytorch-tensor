```python
import torch
```

```python
# Tensor Indexing

batch_size = 10
features = 25
x = torch.rand((batch_size, features))

print(x[0])

print(x[0].shape) # x[0,:]

print(x[:, 0].shape)

print(x[2, 0:10])

x[0, 0] = 100
```

```python
# Fancy indexing
x = torch.arange(10)
indices = [2, 5, 8]
print(x[indices])
```

```python
x = torch.rand((3, 5))
print(x)
rows = torch.tensor([1,0])
cols = torch.tensor([4,0])
# will pick out (1,4) 2nd row, 5th col && (0,0) 1st row, 1st col
print(x[rows, cols])
print(x[rows, cols].shape)
```

```python
# More advanced indexing
x = torch.arange(10)
print(x[(x < 2) | (x > 8)])
print(x[(x < 2) & (x > 8)])
print(x[(x > 2) & (x > 8)])
```

```python
print(x[x.remainder(2) == 0]) # even elements (modulus)
```

```python
# Useful operations
print(torch.where(x > 5, x, x*2)) # if x > 5, x else x*2
```

```python
print(torch.tensor([0,0,1,2,2,3,4]).unique())
```

```python
print(x.ndimension())
```

```python
print(x.numel()) # number of elements in x
```
