```python
import torch
```

```python
# Tensor Math & Comparison Operations

x = torch.tensor([1, 2, 3])
y = torch.tensor([9, 8, 7])
```

```python
# Addition (method 1)
z1 = torch.empty(3)
torch.add(x, y, out=z1)

# method 2
z2 = torch.add(x, y)

# method 3 (preferred)
z = x + y

print(z)
```

```python
# Subtraction
z = x - y

print(z)
```

```python
# Division
z = torch.true_divide(x, y)
print(z)
```

```python
# inplace operations
t = torch.zeros(3)
t.add_(x) # method 1
t += x # method 2
print(t)
```

```python
# Exponentiation
z = x.pow(2) # method 1
z = x ** 2
print(z)
```

```python
# Simple comparison
z = x > 0
z = x < 0
print(z)
```

```python
# Matrix Multiplication
x1 = torch.rand((2, 5))
x2 = torch.rand((5, 3))

x3 = torch.mm(x1, x2) # 2x3 # method 1
x3 = x1.mm(x2) # method 2

print(x3)
```

```python
# Matrix Exponentiation
matrix_exp = torch.rand(5, 5)
matrix_exp.matrix_power(3)
print(matrix_exp)
```

```python
# element wise mult.
z = x * y
print(z)
```

```python
# dot product (sum of element wise mult.)
z = torch.dot(x, y) 
print(z)
```

```python
# Batch Matrix Multiplication
batch = 32
n = 10
m = 20
p = 30

tensor1 = torch.rand((batch, n, m))
tensor2 = torch.rand((batch, m, p))
out_bmm = torch.bmm(tensor1, tensor2) # (batch, n, p)
```

```python
# Example of Broadcasting
x1 = torch.rand((5, 5))
x2 = torch.rand((1, 5)) 

# broadcasting happens to automatically expand x2 to match x1 for the subtraction to happen
z = x1 - x2 
z = x1 ** x2
print(z)
```

```python
# Other useful tensor operations
sum_x = torch.sum(x, dim=0)
print(sum_x)
```

```python
values, indices = torch.max(x, dim=0)
print(values, indices)

values, indices = torch.min(x, dim=0)
print(values, indices)
```

```python
abs_x = torch.abs(x)
print(abs_x)
```

```python
z = torch.argmax(x, dim=0) # index of the max
print(z)

z = torch.argmin(x, dim=0) # index of the min
print(z)
```

```python
mean_x = torch.mean(x.float(), dim=0) # to get mean, pytorch requires a float
print(mean_x)
```

```python
z = torch.eq(x, y) # vals in x and y are equal?
print(z)
```

```python
sorted_y, indices = torch.sort(y, dim=0, descending=False)
print(sorted_y, indices)
```

```python
z = torch.clamp(x, min=2, max=2) # set min and max
print(z)
```

```python
x = torch.tensor([1,0,1,1,1], dtype=torch.bool)
z = torch.any(x)
print(z)
z = torch.all(x)
print(z)
```
